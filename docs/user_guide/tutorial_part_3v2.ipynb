```python
import ross as rs
import ross.stochastic as srs
from ross.probe import Probe
import numpy as np

# Make sure the default renderer is set to 'notebook' for inline plots in Jupyter
import plotly.io as pio
import plotly.graph_objects as go

pio.renderers.default = "notebook"
```

```python
from ross.materials import steel
```

```python
L = 0.25
N = 5  # Number of elements
l_list = [L for _ in range(N)]
shaft_elements = [
    rs.ShaftElement(
        L=l,
        idl=0.0,
        odl=0.05,
        material=steel,
        shear_effects=True,
        rotary_inertia=True,
        gyroscopic=True,
        is_random=["odl"],
    )
    for l in l_list
]
```

```python
m = 32.5
Id = 0.175
Ip = 0.325
#is_random = ["m", "Id", "Ip"]

disk0 = rs.DiskElement(n=2, m=m, Id=Id, Ip=Ip)
#list(iter(disk0))
```

```python
i_d = 0.055
o_d = 0.37
disk1 = rs.DiskElement.from_geometry(
    n=3,
    material=steel,
    width=0.07,
    i_d=i_d,
    o_d=o_d,
    is_random=["i_d", "o_d"],
)
list(iter(disk1))
```

```python
# Building bearing elements and matching their coefficients.

kxx = 1.5e6
cxx = 1.5e3
brg0 = rs.BearingElement(
    n=0,
    kxx=kxx,
    cxx=cxx,
    is_random=["kxx", "cxx"],
)

brg1 = rs.BearingElement(
    n=5,
    kxx=kxx,
    cxx=cxx,
    is_random=["kxx", "cxx"],
)
list(iter(brg0))
```

```python
rotor1 = rs.Rotor(
    shaft_elements,
    [disk0, disk1],
    [brg0, brg1],
)
```

```python
elemen = ['bearing_elements','disk_elements','shaft_elements','shaft_materials']
param = [['kxx','kyy','cxx','cyy'],['Id','Ip','m'],['odl'],['E']]
```

```python
rotorstoch = srs.ST_Make_it_Stochastic(rotor1, elemen, param,store=True)
```

```python
samples = 31
speed_range = np.linspace(0, 500, samples)

# NMC is the number of different values
campbell = rotorstoch.run_stCampbell(NMC=5, speed_range=speed_range)
```

```python
campbell.plot_nat_freq(conf_interval=[90])
```

```python
speed_range = np.linspace(0, 500, 301)
inp = 3 * rotor1.number_dof + 1
out = 2 * rotor1.number_dof + 1
freqresp = rotor1.run_freq_response(inp, out, speed_range)
```

```python
fig2 = freqresp.plot(conf_interval=[90], mag_kwargs=dict(yaxis=dict(type="log")))
fig2.show(renderer="notebook")
```
