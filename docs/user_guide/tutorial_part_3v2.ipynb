```python
import ross.ross as rs
import ross.stochastic as srs
from ross.probe import Probe
import numpy as np

# Make sure the default renderer is set to 'notebook' for inline plots in Jupyter
import plotly.io as pio
import plotly.graph_objects as go

pio.renderers.default = "notebook"
```

```python
from ross.stochastic.st_make_it_stochastic import ST_Make_it_Stochastic
```

```python
# material
Steel = rs.Material(name ="Steel", rho =7800 , E=2e11 , Poisson =0.3)

# shaft
n = 13
shaft_elem = [
rs.ShaftElement(
L=0.1 ,
idl =0,
odl =0.05 ,
material =Steel ,
shear_effects =True ,
rotary_inertia =True ,
gyroscopic =True ,
)
for _ in range (n)
]

# disks
disk1 = rs.DiskElement.from_geometry(
n=2,
material =Steel ,
width =0.05 ,
i_d =0.05 ,
o_d =0.12
)

disk2 = rs.DiskElement.from_geometry(
n=5,
material =Steel ,
width =0.05 ,
i_d =0.05 ,
o_d =0.2
)

disk3 = rs.DiskElement.from_geometry(
n=10 ,
material =Steel ,
width =0.06 ,
i_d =0.05 ,
o_d =0.2
)

# stiffness values
kx = 5e7
ky = 7e7

# damping values
cx = 5e3
cy = 7e3

# bearings
bearing1 = rs. BearingElement (n=0, kxx=kx , kyy=ky , cxx=cx , cyy=cy )

bearing2 = rs. BearingElement (n=13 , kxx=kx , kyy =ky , cxx=cx , cyy=cy)

disks = [disk1 , disk2 , disk3 ]
bearing = [ bearing1 , bearing2 ]

# rotor model
rotor = rs.Rotor(shaft_elem, disks, bearing)
```
